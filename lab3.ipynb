```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('2.jpg', 0)
image = image.astype(np.uint8)

plt.figure(figsize=(6, 6))
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.show()
```

### Negative

```python
neg = 255 - image
plt.imshow(neg, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Logarithmic Transformation

```python
image = image.astype(np.float32)
```

```python
c = 255 / np.log1p(np.max(image))
lt = c * np.log1p(image)
lt = np.clip(lt, 0, 255)
lt = lt.astype(np.uint8)
```

```python
plt.imshow(lt, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Power-Law Transformation

```python
g = 0.7
normalized = image / 255.0
pl = np.power(normalized, g)
pl = np.uint8(pl * 255)
```

```python
plt.imshow(lt, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Contrast Stretching

```python
rmin = np.min(image)
rmax = np.max(image)
cs = (image - rmin) * (255 / (rmax - rmin))
cs = np.uint8(cs)
```

```python
plt.imshow(cs, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Thresholding

```python
tval = 128
t = np.where(image > tval, 255, 0)
```

```python
plt.imshow(t, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Bit-Plane Slicing

```python
image_uint8 = image.astype(np.uint8)
bit_planes = []
for i in range(8):
    bit_plane = (image_uint8 >> i) & 1
    bit_planes.append(bit_plane * 255)
```

```python
plt.figure(figsize=(12, 6))
for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_planes[i], cmap='gray')
    plt.title(f'{i}')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

### Gray-Level Slicing

```python
a, b = 100, 180
gs1 = image.copy()
gs1[(image >= a) & (image <= b)] = 255

gs2 = np.where((image >= a) & (image <= b), 255, 0).astype(np.uint8)
```

```python
plt.imshow(gs1, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
plt.imshow(gs2, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```
