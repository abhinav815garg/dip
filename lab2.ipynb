### Nearest Neighbour Interpolation

```python
import matplotlib.image as img
import numpy as np
```

```python
original = img.imread("1.jpg")
print(original.shape)
```

```python
x, y = original.shape[:2]
print(x, y)
```

```python
scale = 1.2
x_n, y_n = int(x * scale), int(y * scale)
print(x_n, y_n)
```

```python
x_scale, y_scale = x / x_n, y / y_n
print(x_scale, y_scale)
```

```python
new_image = np.zeros([x_n, y_n, 3], dtype=np.uint8)
for i in range(x_n):
    for j in range(y_n):
        src_x = int(i * x_scale)
        src_y = int(j * y_scale)
        new_image[i, j] = original[src_x, src_y]
```

```python
img.imsave("scaled.jpg", new_image)
```

### Gray Level Resolution

```python
from statistics import mean
```

```python
original = img.imread("3.jpg")
print(original.shape)
x, y = original.shape[:2]
print(x, y)
```

```python
gray = original[:, :, 0]
print(gray.shape)
img.imsave("temp.jpg", gray, cmap="gray")
```

```python
gray = gray / 255.0
```

```python
L = 8
gray_res = np.zeros((x, y))
for i in range(x):
    for j in range(y):
        gray_res[i, j] = np.floor(gray[i, j] * (L - 1)) / (L - 1)
```

```python
img.imsave("grayedLevelResolution.jpg", gray_res, cmap="gray")
```
