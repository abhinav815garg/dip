```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
import cv2
```

```python
img = cv2.imread("1.jpg")
print(img.shape)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(gray, cmap="gray")
plt.axis("off")
```

```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(rgb)
plt.axis("off")
```

```python
gray_img_2 = cv2.imread("2.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(gray_img_2, cmap="gray")
plt.axis("off")
print("Shape: ", gray_img_2.shape)
```

```python
bgr_from_gray = cv2.cvtColor(gray_img_2, cv2.COLOR_GRAY2BGR)
plt.imshow(bgr_from_gray)
plt.axis("off")
print("Shape: ", bgr_from_gray.shape)
```

```python
rgb_from_gray = cv2.cvtColor(gray_img_2, cv2.COLOR_GRAY2RGB)
plt.imshow(rgb_from_gray)
plt.axis("off")
print("Shape: ", rgb_from_gray.shape)
```

```python
new_width = 600
new_height = 500

resized_image = cv2.resize(rgb_from_gray, (new_width, new_height))

cv2.imshow("Resized Image", resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
scale_down = 0.5  
scale_up = 1.5    
 
resized_down = cv2.resize(img, None, fx=scale_down, fy=scale_down, interpolation=cv2.INTER_LINEAR)
 
resized_up = cv2.resize(img, None, fx=scale_up, fy=scale_up, interpolation=cv2.INTER_LINEAR)
 
cv2.imshow("Original Image", img)
cv2.imshow("Scaled Down Image", resized_down)
cv2.imshow("Scaled Up Image", resized_up)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
